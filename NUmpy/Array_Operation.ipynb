Deep And Shallow copy


```python
a=np.arange(1,21)
a
```

Shallow Copy


```python
slice=a[0:10] #it create shallow copy of the original array, so any changes made to the slice will also affect the original array. This is because both the slice and the original array share the same memory location. If you want to create a new array that is independent of the original array, you can use the copy() method to create a deep copy of the slice.
slice
```

```python
a    #on slicing a shallow copy is created, so if we change the slice it will not affect the original array
```

Deep Copy

```python
a
b=a  #Create a new variable b and assign it the value of a. This creates a reference to the same array in memory, so changes made to b will affect a as well.
```

```python
b[0]=100
b
```

```python
a  #changes made to b will also affect a, since they are both referencing the same array in memory.
```

```python
#for creating shaLLOW COPY OF A USE copy() method
c=a.copy()
c[0]=1000
c
```

```python
a  #changes made to c will not affect a, since c is a separate copy of the array in memory.
```

Advance Array Manipulation

Stacking array


```python
#Two Matrix can't be stacked if they have different number of columns. The number of columns in both matrices must be the same for them to be stacked vertically. If the number of columns is different, it will result in a ValueError.
```

```python
a=np.array([1,2,3,4,5])
b=np.array([6,7,8,9,10])
```

```python
np.vstack((a,b))  #vertical stacking of two arrays
```

```python
np.hstack((a,b))  #horizontal stacking of two arrays
```

```python
np.column_stack((a,b))  #column stacking of two arrays
```

Splitting Array 



```python
d=np.arange(16).reshape(4,4)
d
```

```python
np.hsplit(d,2)  #split the array into 2 equal parts horizontally
```

```python
np.hsplit(d,4)  #split the array into 4 equal parts horizontally
```

```python
np.vsplit(d,2)  #split the array into 2 equal parts vertically
```

```python
z=np.vsplit(d,4)  #split the array into 4 equal parts vertically  
```

```python
for i in z:    #may be z have splited array but it is stored in form of matrix so we have to iterate over it to get the individual array
    print(i)
```
