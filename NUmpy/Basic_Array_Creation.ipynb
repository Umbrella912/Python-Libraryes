Array generation function


```python
np.arange(1,11)
```

```python
arr=np.zeros(3) #for 1D array
arr1=np.zeros((3,4)) #for 2D array  (rows,columns)
print(arr)
print(arr1) 
```

Random Generation function


```python
np.random.rand(5) #random numbers between 0 and 1     #Normalization of data is done using this function
```

```python
np.random.randn(5) #random numbers from standard normal distribution  #standard normal distribution has mean=0 and std=1
```

```python
np.random.randint(10,20,11) #random integers between 10 and 20 (20 is exclusive) and 11 is the number of elements
```

Reshape and Resize


```python
arr=np.arange(1,11)
```

```python
arr.reshape(5,2) #reshape the array to 5 rows and 2 columns

#Remember no of elements should be same in original and reshaped array
#tere for select the number of rows and columns such that their product is equal to the number of elements in the original array
```

```python
arr=arr.reshape(5,2) #Assigning the reshaped array to original array is called resizing the array. Now arr is a 2D array with 5 rows and 2 columns
```
