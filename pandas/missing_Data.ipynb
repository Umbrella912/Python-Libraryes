```python
import numpy as np
import pandas as pd
```

# Finding Missing Data
 

```python
data={
    'A':[1,2,3,np.nan,5],
    'b':[np.nan,2,3,np.nan,5],
    'c':[1,np.nan,3,4,5],
    'd':[1,2,3,4,5]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum() #Null Value in each columns
```

```python
df.isna().any()
```

# Removing Missing Data

```python
df
```

```python
df.dropna() #drop the row contaning null value in any columns
```

```python
df.dropna(thresh=3) # it will remove every row that not contaning at least 3 not null values 
```

```python
df.dropna(thresh=5)
```

# Filling The missing data


```python
df
```

```python
df.fillna(0)
```

```python
values={'A':10,'b':20,'c':30,'d':40}
df.fillna(value=values)
```

```python
df.fillna(df.mean())
```
