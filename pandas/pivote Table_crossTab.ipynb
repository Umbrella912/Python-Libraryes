```python
import pandas as pd
import numpy as np
```

```python


# Set seed for reproducible random data
np.random.seed(42)

# Generate 100 rows of categorical data matching typical tutorial sets
n_rows = 100
dates = pd.date_range(start="2026-01-01", periods=4, freq="D").strftime("%Y-%m-%d").tolist() * 25
regions = np.random.choice(["North", "South", "East", "West"], size=n_rows)
products = np.random.choice(["Laptop", "Smartphone", "Tablet", "Headphones"], size=n_rows)
sales = np.random.randint(100, 1500, size=n_rows)
quantity = np.random.randint(1, 10, size=n_rows)

# Create the data frame
df = pd.DataFrame({
    "Date": dates,
    "Region": regions,
    "Product": products,
    "Sales": sales,
    "Quantity": quantity
})

# Display the first 10 rows
print("--- Raw DataFrame (First 10 Rows) ---")
print(df.head(10))
```

# Pivote Table


A pivot table is used to transform messy, row-by-row data into a clean, summarized summary table without changing originalm table


used in creation of heatmap

```python

pd.pivot_table(df,values='Sales',index='Region',columns='Product') # by default povit table give mean for ex
#677.5 is mean of hadephone in east region
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product', aggfunc='median') 
```

```python
pivot=pd.pivot_table(df,values=['Sales','Quantity'],index='Region',columns='Product') #values must be quantative(Numerical) other wiase agg function will fail
pivot
```
