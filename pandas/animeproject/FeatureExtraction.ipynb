```python
def extract_episode(text):
    check=False
    data=""
    for i in text:
        if i==")":
            check=False
            return data 
        if check ==True:
            data+=i             
        if i=='(':
            check=True
         
```

```python
df['Episodes']=df['Title'].apply(extract_episode)
df.head()
```

```python
df['Episodes']=df['Episodes'].str.replace(" eps","")
df['Episodes']=df['Episodes'].astype(int)
df.head()
```

Make new column of Total time

```python
def extract_date(text):
    check=False
    data=""
    for i in range(len(text)):
        if text[i]==')':
            for j in range(i+1,i+20):
                data+= text[j]
            return data 
         
```

```python
df['Total time']=df['Title'].apply(extract_date)
df.head()
```

Make New Column of total month

```python


def total_months(s, inclusive=True):
    try:
        start, end = [x.strip() for x in s.split('-')]
        start = pd.to_datetime(start, format='%b %Y')
        # handle "Present"/"Current" as today's date
        if end.lower() in ('present', 'current', 'till date'):
            end = pd.Timestamp.today()
        else:
            end = pd.to_datetime(end, format='%b %Y')
        months = (end.year - start.year) * 12 + (end.month - start.month)
        return months + 1 if inclusive else months
    except Exception:
        return None  # malformed or missing values

```

```python
df['total_months'] = df['Total time'].apply(total_months)
df.head()
```
