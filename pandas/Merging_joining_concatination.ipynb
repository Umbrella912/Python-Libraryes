# MErging 2 Data frame



1. pd.merge() 

(Database-Style Joining)Use merge when you want to combine DataFrames based on common values in specific columns, exactly like an SQL JOIN.

Best for: Connecting tables with shared ID columns (e.g., matching employee_id in a names table to employee_id in a salaries table).

Flexibility: Very high. Column names do not have to match exactly between tables (left_on and right_on handle mismatches).


```python
data1 = {
    "employee_id": [101, 102, 103, 104],
    "name": ["Alice", "Bob", "Charlie", "David"]
}
data2 = {
    "employee_id": [101, 102, 103, 105],
    "salary": [50000, 60000, 55000, 70000]
}
```

```python
a=pd.DataFrame(data1)
print(a)
b=pd.DataFrame(data2)
print(b)
```

```python
pd.merge(a,b,on='employee_id') # merge always need one common colum between table to be merge
```

```python
pd.merge(a,b,on="employee_id",how="inner") #it will only merge those column that have common value in common column
```

```python
pd.merge(a,b,on="employee_id",how="outer") # it will merge every row and fill null where values are not avaliable
```

```python
pd.merge(a,b,on="employee_id",how="left") #it will merge all rows of left side table
```

```python
pd.merge(a,b,on="employee_id",how="right")  #it will merge all rows from right side
```

# Conctination of 2 Dataframe

* pd.concat()  

(Stacking Data)Use concat when you want to glue or stack multiple DataFrames together along an axis (either on top of each other or side-by-side).

Best for: Combining data that has been split up (e.g., stacking monthly sales files like January, February, and March into one yearly dataset).

Flexibility: Can combine more than two DataFrames at once.

```python
data_jan = {
    "product": ["Laptop", "Mouse", "Keyboard"],
    "units_sold": [50, 120, 80]
}
data_feb = {
    "product": ["Laptop", "Mouse", "Monitor"],
    "units_sold": [45, 110, 35]
}
sales1=pd.DataFrame(data_jan)
sales2=pd.DataFrame(data_feb)
```

```python
sales1
```

```python
sales2
```

```python
pd.concat([sales1,sales1])
```

```python
pd.concat([sales1,sales2],axis=1)
```

# Joining 2 DataFrame

3. DataFrame.join() (Index-Based Combining) 

Use join as a convenient shorthand to combine DataFrames based on their row indexes instead of their columns

Best for: Fast lookup combinations where your unique identifiers are already set as the DataFrame index.

Flexibility: By default, it looks strictly at row indexes, though you can configure it to join a left column onto a right index.

```python
df_users = pd.DataFrame(
    {"name": ["Alice", "Bob", "Charlie"]},
    index=[101, 102, 103]
)
df_activity = pd.DataFrame(
    {"login_count": [12, 45, 8]},
    index=[101, 102, 104]
)
df1=pd.DataFrame(df_users)
df2=pd.DataFrame(df_activity)
```

```python
df1
```

```python
df2
```

```python
df1.join(df2) # left join
```

```python
df1.join(df2,how='left')
```

```python
df1.join(df2,how='right')
```

```python
df1.join(df2,how='inner')
```

```python
df1.join(df2,how='outer')
```
