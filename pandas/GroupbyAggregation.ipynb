# GroupBy

```python
data = {
    "date": ["2026-09-25", "2026-09-25", "2026-09-26", "2026-09-26", "2026-09-27", "2026-09-27"],
    "store": ["Store A", "Store B", "Store A", "Store B", "Store A", "Store B"],
    "category": ["Electronics", "Clothing", "Electronics", "Electronics", "Clothing", "Clothing"],
    "sales": [1200.50, 450.00, 800.00, 1500.00, 300.00, 650.75],
    "quantity": [3, 15, 2, 4, 10, 22]
}
df=pd.DataFrame(data)
```

```python
df
```

Group by Category and calculate the sum of sales

```python
c=df.groupby("category") #igroupby will generate object , that you can't access in general way
for i,v in c:
    print(i)
    print(v)
```

```python
df.groupby('category')['sales'].sum()
```

```python
df.groupby('category')[['sales','quantity']].sum()
```

```python
#Group by store and do sum of sales
df.groupby("store")['sales'].sum()
```

```python
#groupby Multiple column
#group by category and store 
df.groupby(["category","store"])['sales'].sum()
```

```python
df.groupby(["category","store"])[['sales','quantity']].sum()
```

```python
df.groupby(["category","store"])[['sales','quantity']].agg(['sum','mean','max','min'])
```
